```python
import pandas as pd

# Simulate data extracted from multiple mock stores
def extract_store_data(store_id):
    try:
        # Mocking extraction logic (in real life, this is your BeautifulSoup/Playwright code)
        if store_id == 99:
            raise Exception("Connection timeout") # Simulating a failed store

        data = {
            'Store_ID': [store_id, store_id],
            'Product': ['Item A', 'Item B'],
            'Units_Sold': [10, 25],
            'Unit_Price': [15.0, 40.0]
        }
        return pd.DataFrame(data), None
    except Exception as e:
        return None, str(error := e)

# Process 3 stores with robust independent try/except error handling
all_store_data = []
error_logs = []

store_ids = [1, 99, 3] # Store 99 will trigger our simulated error

for s_id in store_ids:
    df, err = extract_store_data(s_id)
    if err:
        error_logs.append(f"Store {s_id} failed: {err}")
    else:
        all_store_data.append(df)

# Consolidate successful extractions using Pandas
if all_store_data:
    final_df = pd.concat(all_store_data, ignore_index=True)

    # Calculate Commission, Cost Price, and Net Profit
    final_df['Total_Revenue'] = final_df['Units_Sold'] * final_df['Unit_Price']
    final_df['Cost_Price'] = final_df['Total_Revenue'] * 0.6  # Example rule: 60% cost
    final_df['Commission'] = final_df['Total_Revenue'] * 0.1  # Example rule: 10% commission
    final_df['Net_Profit'] = final_df['Total_Revenue'] - final_df['Cost_Price'] - final_df['Commission']

    print("--- Consolidated Sales & Profit Report ---")
    display(final_df)

print("\n--- Error Log Summary ---")
for log in error_logs:
    print(log)
```

```python
# Export the consolidated report to an Excel sheet
output_file = 'daily_sales_report.xlsx'
final_df.to_excel(output_file, index=False)

print(f"Report successfully saved to {output_file}")
```
